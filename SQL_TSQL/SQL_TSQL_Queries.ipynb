# Create Tables

## PROGRAMMER Table

```sql
CREATE TABLE IF NOT EXISTS PROGRAMMER(
  PNAME VARCHAR (50) NOT NULL,
  DOB DATE NOT NULL,
  DOJ DATE NOT NULL,
  GENDER CHAR(1) NOT NULL,
  PROF1 VARCHAR (50) NOT NULL,
  PROF2 VARCHAR (50) NOT NULL,
  SALARY INT NOT NULL,
  PRIMARY KEY (PNAME)
);
```

## STUDIES Table

```sql
CREATE TABLE IF NOT EXISTS STUDIES(
  PNAME VARCHAR (50) NOT NULL, 
  INSTITUTE VARCHAR (50) NOT NULL,
  COURSE VARCHAR (50) NOT NULL,
  COURSE_FEE INT NOT NULL, 
  FOREIGN KEY(PNAME) REFERENCES PROGRAMMER(PNAME)
);
```

## SOFTWARE Table

```sql
CREATE TABLE IF NOT EXISTS SOFTWARE(
  PNAME VARCHAR (50) NOT NULL,
  TITLE VARCHAR (50) NOT NULL,
  DEVELOPIN VARCHAR(50) NOT NULL,
  SCOST FLOAT NOT NULL,
  DCOST FLOAT NOT NULL,
  SOLD INT NOT NULL,
  FOREIGN KEY(PNAME) REFERENCES PROGRAMMER(PNAME)
);
```

### 1. Find out the selling cost Average for packages developed in Pascal. 

```sql
SELECT DEVELOPIN, AVG(SCOST) AS AVERAGE_SELLING_COST FROM SOFTWARE WHERE DEVELOPIN = 'PASCAL' GROUP BY DEVELOPIN; 
```

### 21. Display the details of the software developed in DBASE by Male Programmers, who belong to the institute in which most number of Programmers studied.  

SUBQUERY:

### 22. Display the details of the software Developed by the male programmers Born before 1965 and female programmers born after 1975. 

```sql
SELECT S.PNAME, P.GENDER, P.DOB, S.TITLE, S.DEVELOPIN, S.SCOST, S.DCOST, S.SOLD FROM SOFTWARE S INNER JOIN PROGRAMMER P ON S.PNAME = P.PNAME WHERE (
  P.GENDER = 'M' AND YEAR(P.DOB) > 1965) OR (P.GENDER = 'F' AND YEAR(P.DOB) > 1975);
```

### 23. How many packages were developed by the female programmers earning more than the highest paid male programmer? 

```sql
SELECT COUNT(S.TITLE) AS NUMBER_OF_PACKAGES FROM SOFTWARE S INNER JOIN PROGRAMMER P ON S.PNAME = P.PNAME WHERE P.GENDER='F' AND P.SALARY>(
  SELECT MAX(SALARY) FROM PROGRAMMER WHERE GENDER='M');
```

### 24. How many packages were developed by the person who developed the cheapest package, where did HE/SHE study? 


```sql
SELECT COUNT(*) AS NUMBER_OF_PACKAGES, ST.INSTITUTE FROM SOFTWARE S INNER JOIN STUDIES ST ON S.PNAME = ST.PNAME WHERE S.PNAME IN (
  SELECT PNAME FROM SOFTWARE ORDER BY DCOST LIMIT 1) GROUP BY ST.INSTITUTE;
```

### 25. Display the details of the software that has developed in the language which is neither the first nor the second proficiency of the programmers.

### 26. Who are the programmers who joined on the same day?  

```sql
SELECT P1.PNAME, P1.DOJ FROM PROGRAMMER P1 WHERE DOJ IN (
  SELECT DOJ FROM PROGRAMMER P2 WHERE P1.PNAME != P2.PNAME AND P1.DOJ = P2.DOJ);
```

### 27. Who are the programmers who have the same Prof2?  

```sql
SELECT DISTINCT P1.PNAME, P1.PROF2 FROM PROGRAMMER P1 INNER JOIN PROGRAMMER P2 ON  P1.PROF2 = P2.PROF2 WHERE P1.PNAME != P2.PNAME;
```

### 28. Display the total sales value of the software, institute wise.
